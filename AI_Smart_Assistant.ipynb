```python
print("🤖 AI Smart Assistant (Agent-Based)")
print("Type 'exit' to quit\n")

def detect_task(text):
    text = text.lower()
    if "summarize" in text:
        return "summarize"
    elif "search" in text:
        return "search"
    elif "email" in text:
        return "email"
    else:
        return "chat"

def summarize(text):
    words = text.split()
    return " ".join(words[:20]) + "..."  # simple summary

def generate_email(topic):
    return f"""
Subject: Regarding {topic}

Dear Sir/Madam,

I hope you are doing well. I am writing regarding {topic}.
I would like to express my interest and request further information.

Thank you for your time.

Best regards,
Your Name
"""

def chat_response(text):
    return f"I understand your query: '{text}'. This can be enhanced using AI models."

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Goodbye!")
        break

    task = detect_task(user_input)

    if task == "summarize":
        print("📄 Summary:", summarize(user_input))

    elif task == "email":
        print("✉️ Email:", generate_email(user_input))

    elif task == "search":
        print("🌐 Search Result: Showing top results for:", user_input)

    else:
        print("💬 Response:", chat_response(user_input))
```
